# NumPy 소개
NumPy(Numerical Python)는 파이썬 과학 계산의 핵심 라이브러리입니다. 다차원 배열 객체 <mark>ndarray</mark> 와 이를 다루는 빠른 연산 함수를 제공하며.
Pandas, SciPy, scikit-learn, matplotlib 등 거의 모든 데이터 분석 도구가 NumPy 위에서 동작합니다.

## 왜 NumPy 인가
- **속도**: 내부가 C로 구현되어 순수 파이썬 반복문보다 수십~수백 배 빠르다.
- **벡터화**: 반복문 없이 배열 전체에 연산을 한 번에 적용한다.
- **메모리 효율**: 같은 자료형을 연속된 메모리에 저장해 오버헤드가 적다.
- **생태계**: 데이터 과학 라이브러리의 공통 데이터 구조로서 자리잡아 있다.

```python
# Import 이후 설치 확인 (버전 확인)
import numpy as np # convention: numpy는 대개 np를 별칭으로 붙여서 임포트 함.
print(np.__version__)
```

```python
# 파이썬 리스트 vs NumPy 배열 테스트
# 리스트는 원소별로 반복해야 하지만, 배열은 벡터화 연산이 가능하다
# 연산 방식도 다름.
py_list = [1, 2, 3, 4]
arr = np.array([1, 2, 3, 4])
print("Python List *2:", py_list * 2) # 반복 (복제)
print("NumPy Array *2:", arr * 2) # 원소별 곱셈
```

```python
# 속도 비교: 100만개 원소의 제곱 합
import time
n = 1_000_000
data = list(range(n))
arr = np.arange(n) # arange는 0부터 n까지 차례로 나열한 배열을 반환.
#print(arr)

t0 = time.perf_counter()
s1 = sum(x*x for x in data) # 순회하며 제곱을 더함.
t1 = time.perf_counter()
s2 = int((arr*arr).sum()) # 원소 별 곱 -> 총합
t2 = time.perf_counter()
print("결과 동일: ", s1 == s2)
print(f"순수 파이썬 실행 시간: {t1-t0}")
print(f"NumPy 배열 연산 실행 시간: {t2-t1}")
```
